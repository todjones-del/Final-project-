## 💵 Milestone 6: Work Out the Total

**Your job:** Write a function named `order_total` that adds up the prices of every item in the order, then adds sales tax.

**Steps:**

2. Loop over the `order` list.

````python
# TODO: Write a comment here that says what order_total does.
def order_total(order, menu, tax_rate=0.08):
    subtotal = 0
    for item in order:
        subtotal += get_price(menu,item)
    total = subtotal * (1 + tax_rate)
    return round(total, 2)
````

### Part B: Check the quantity

**Your job:** Write a function named `parse_quantity`. It takes text and returns a whole number greater than 0. If the text is not a valid quantity, it returns `None`.

**Steps:**
1. In `try`, convert the text with `int(text)`.
2. If the text is not a whole number, `int` raises a `ValueError`. Catch it and return `None`.
3. After the `try`, return the number if it is greater than 0. Otherwise return `None`.

### Part C: Keep asking

**Your job:** Finish `ask_quantity`. It keeps asking until the customer types a valid quantity.

1. Fill in the blank with the words `is not`.

**Expected output after the checks:** `PASS` for every line.

````python
# Cleans leading and trailing whitespace from a name and capitalizes each word.
def clean_name(text):
    return text.strip().title()

# Converts a string to an integer, returning it if it's greater than 0, otherwise returns None.
def parse_quantity(text):
    try:
        number = int(text)
    except ValueError:
        return None
    if number > 0:
        return number
    else:
        return None

# Continually asks the user for a quantity until they enter a valid whole number greater than 0.
def ask_quantity():
    while True:
        text = input('How many? ')
        quantity = parse_quantity(text)
        if quantity is not None:
            return quantity
        print('Please type a whole number greater than 0.')
````

### Part A: add_points

**Your job:** Write a function named `add_points` that adds points to a customer and returns their new total. New customers are added to the dictionary.

**Steps:**

2. If the name is **in** `customers`, add the points with `+=`.

### Part B: get_tier

**Your job:** Write a function named `get_tier` that returns a tier name from a points total.

- 200 points or more: `'Gold'`
- 100 points or more: `'Silver'`
- Anything less: `'Bronze'`

**Expected output after the checks:** `PASS` for every line.

````python
# Adds loyalty points to a customer's total (1 point per whole dollar spent) and returns their new total.
def add_points(customers, name, total):
    points = int(total)
    if name in customers:
        customers[name] += points
    else:
        customers[name] = points
    return customers[name]

# Determines the loyalty tier ('Gold', 'Silver', or 'Bronze') based on a customer's accumulated points.
def get_tier(points):
    if points >= 200:
        return 'Gold'
    elif points >= 100:
        return 'Silver'
    else:
        return 'Bronze'
````

## 🧾 Milestone 13: Print a Receipt

### Part A: Count the items

**Your job:** Write a function named `count_items` that turns an order list into a dictionary of counts. For example, `['Latte', 'Muffin', 'Latte']` becomes `{'Latte': 2, 'Muffin': 1}`.

**Steps:**
1. Start with an empty dictionary named `counts`.
2. Loop over the order.
3. If the item is already a key, add 1 to its count with `+=`.
4. Otherwise, set its count to `1`.
5. Return `counts`.

### Part B: Print the receipt

**Your job:** Finish `print_receipt`. It uses `count_items` and `order_total`.

**Steps:**
1. Fill in `.items()` to loop over the counts.
2. The line `days, open_hour, close_hour = hours` **unpacks** the tuple into three variables. Fill in the missing name.
3. Fill in the missing parts of the f-strings.

**Expected output for the test call at the bottom:**
```
--- RECEIPT ---
2 x Latte: $9.00
1 x Muffin: $3.25
Total with tax: $13.23
Open Mon-Fri, 7 to 18
```

````python
# Counts the occurrences of each item in the order and returns a dictionary with item counts.
def count_items(order):
    counts = {}
    for item in order:
        if item in counts:
            counts[item] += 1
        else:
            counts[item] = 1
    return counts

# Prints a formatted receipt displaying items, quantities, totals, and cafe operating hours.
def print_receipt(order, menu):
    print('--- RECEIPT ---')
    counts = count_items(order)
    for item, quantity in counts.items():
        line_total = get_price(menu, item) * quantity
        print(f'{quantity} x {item}: ${line_total:.2f}')
    print(f'Total with tax: ${order_total(order, menu):.2f}')
    days, open_hour, close_hour = hours
    print(f'Open {days}, {open_hour} to {close_hour}')

print_receipt(['Latte', 'Muffin', 'Latte'], menu)
````

## 📊 Milestone 14: The Daily Sales Report

At closing time, the owner wants a report. The `sales_log` list holds every finished order, and each order is itself a list of items.

### Part A: daily_report

**Your job:** Write a function named `daily_report` that returns a dictionary with four numbers about the day.

**Steps:**
1. Loop over `sales_log`. Each `sale` is one order.
2. Add the number of items in the sale to `items_sold` with `len(sale)`.
3. Add the sale's total to `revenue` using `order_total(sale, menu)`.
4. Loop over the items in the sale and add each one to the set `unique`. Use the set method `.add`.
5. Return a dictionary with the keys `'orders'`, `'items'`, `'revenue'`, and `'unique_items'`.

### Part B: print_report

**Your job:** Finish `print_report`. Fill in the last key, `'unique_items'`.

**Expected output for the test call at the bottom:**
```
--- DAILY REPORT ---
Orders: 3
Items sold: 4
Revenue: $15.12
Different items sold: 3
```

````python
# Generates a summary report of daily sales, including total orders, items sold, revenue, and a set of unique items.
def daily_report(sales_log, menu):
    items_sold = 0
    revenue = 0
    unique = set()
    for sale in sales_log:
        items_sold += len(sale)
        revenue += order_total(sale, menu)
        for item in sale:
            unique.add(item)
    return {
        'orders': len(sales_log),
        'items': items_sold,
        'revenue': round(revenue, 2),
        'unique_items': unique
    }

# Prints a formatted daily sales report containing orders count, items sold, revenue, and different items.
def print_report(report):
    print('--- DAILY REPORT ---')
    print('Orders:', report['orders'])
    print('Items sold:', report['items'])
    revenue = report['revenue']
    print(f'Revenue: ${revenue:.2f}')
    print('Different items sold:', len(report['unique_items']))

test_log = [['Latte', 'Muffin'], ['Cookie'], ['Latte']]
print_report(daily_report(test_log, menu))
````

`run_cafe` is the program a customer uses. It asks for a name, shows a menu of choices, reads what the customer types, and calls the functions you built. When the customer quits, it prints the daily report.

````python
# The main program loop that manages cafe orders, menu interaction, allergy filtering, payments, and checkout.
def run_cafe():
    my_order = []
    print(f'Welcome to {cafe_name}!')
    customer = clean_name(input('Your name: '))
    while True:
        print()
        print('1) Menu  2) Menu by category  3) Add item  4) Remove item')
        print('5) View order  6) Allergy-safe items  7) Check out  8) Quit')
        choice = input('Choose 1 to 8: ')
        if choice == '1':
            show_menu(menu)
        elif choice == '2':
            show_menu_by_category(menu, categories)
        elif choice == '3':
            name = clean_name(input('Item name: '))
            if get_price(menu, name) is None:
                print('Sorry, we do not have that item.')
            else:
                quantity = ask_quantity()
                for _ in range(quantity):
                    add_to_order(my_order, menu, name)
                print(f'Added {quantity} x {name}.')
        elif choice == '4':
            name = clean_name(input('Item to remove: '))
            if remove_from_order(my_order, name):
                print(f'Removed one {name}.')
            else:
                print('That item is not in your order.')
        elif choice == '5':
            print('Your order:', my_order)
        elif choice == '6':
            allergy_text = input('Your allergies, separated by commas: ')
            my_allergies = set()
            for word in allergy_text.split(','):
                my_allergies.add(word.strip().lower())
            print('Safe for you:', safe_items(menu, my_allergies))
        elif choice == '7':
            print_receipt(my_order, menu)
            points = add_points(customers, customer, order_total(my_order, menu))
            print(f'{customer} now has {points} points ({get_tier(points)}).')
            sales_log.append(my_order)
            my_order = []
        elif choice == '8':
            print('Goodbye!')
            print_report(daily_report(sales_log, menu))
            break
        else:
            print('Please choose a number from 1 to 8.')
````

**Your job:** Write a function named `apply_discount` that lowers a total when the customer gives a valid code.

````python
codes = {'SAVE10': 0.10, 'FALL20': 0.20}

# Applies a discount percentage to the total price if a valid code is provided.
def apply_discount(total, code):
    if code in codes:
        return round(total * (1 - codes[code]), 2)
    else:
        return total
````

### ⭐ Extension B: Best Seller

Only complete this if you chose B.

**Your job:** Write a function named `best_seller` that returns the item that appears most often in an order. If the order is empty, it returns `None`.

**Steps:**
1. Use `count_items(order)` to get the counts dictionary.
2. Loop over `counts.items()`.
3. Keep track of the best item so far. If an item's quantity is greater than `best_count`, it becomes the new best item.
4. Return `best_item`.

**Expected output after the checks:** `PASS` for every line.

````python
# Determines which item is ordered the most by counting items and finding the maximum quantity.
def best_seller(order):
    counts = count_items(order)
    best_item = None
    best_count = 0
    for item, quantity in counts.items():
        if quantity > best_count:
            best_item = item
            best_count = quantity
    return best_item
````

This function runs without an error, but the total is wrong. For `['Latte', 'Muffin']` it should return `7.75` but it returns `4.5`.

**Hint:** A function leaves as soon as it reaches `return`. Look at where the `return` line is indented.

````python
# BUG FIX: The 'return' statement was indented inside the 'for' loop, causing it to exit after the first item. I moved it outside the loop.
def cart_total(order, menu):
    total = 0
    for item in order:
        total += menu[item]
    return total

print(cart_total(['Latte', 'Muffin'], menu))
````

This function should turn an order into numbered lines, like `['1. Latte', '2. Muffin']`. It runs without an error, but the first item is missing.

**Hint:** `range(1, len(order))` starts at index 1, but lists start at index 0. Also check that the numbers shown start at 1.

````python
# BUG FIX: range(1, len(order)) skipped index 0. Changed range to start at 0, and adjusted string numbering to use i + 1.
def list_order(order):
    lines = []
    for i in range(len(order)):
        lines.append(f'{i + 1}. {order[i]}')
    return lines

print(list_order(['Latte', 'Muffin']))
````
